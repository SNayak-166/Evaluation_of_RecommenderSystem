Cell-1: The movies.csv file available in Kaggle requires genre encoding. Since the dataset use pipe-separated genres (e.g., Action|Adventure|Thriller), we process them using MultiLabelBinarizer and hosted the processed file on GitHub repository. The movie.csv file is hosted in GitHub after downloading from kaggle.

```python
import pandas as pd
from sklearn.preprocessing import MultiLabelBinarizer

github_url = "https://raw.githubusercontent.com/SNayak-166/Evaluation_of_RecommenderSystem/refs/heads/othercodes/movies.csv"

# Load original movies file
movies_df = pd.read_csv(github_url)

# Split the pipe-separated genres into lists
movies_df['genre_list'] = movies_df['genres'].apply(lambda x: x.split('|'))

# Apply MultiLabelBinarizer
mlb = MultiLabelBinarizer()
genre_encoded = pd.DataFrame(mlb.fit_transform(movies_df['genre_list']), columns=mlb.classes_, index=movies_df.index)

# Combine with movieId, title, and genres
processed_movies_df = pd.concat([movies_df[['movieId', 'title', 'genres']], genre_encoded], axis=1)

# Save and upload to your GitHub repo
processed_movies_df.to_csv('processed_movies.csv', index=False)
print("processed_movies.csv is already available in the reposatoray and pipeline.ipynb access it directly")
```
